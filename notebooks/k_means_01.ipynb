```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 12))
sns.heatmap(df.corr(), cmap='coolwarm', annot=False)
plt.title("Correlation Matrix of All Clustering Features")
plt.show()
```

```python
import numpy as np

corr_matrix = df.corr().abs()

threshold = 0.80
high_corr_pairs = []

cols = corr_matrix.columns

for i in range(len(cols)):
    for j in range(i+1, len(cols)):
        corr_value = corr_matrix.iloc[i, j]
        if corr_value >= threshold:
            high_corr_pairs.append((cols[i], cols[j], corr_value))

# Print
print("🔥 Highly Correlated Pairs (>|0.8|):")
for col1, col2, corr in high_corr_pairs:
    print(f"{col1}  —  {col2}   =  {corr:.3f}")
```

A correlation threshold of 0.8 was used to detect redundant features.
Three highly correlated pairs were found:

NumOwned – NumUserRatings (0.985)

NumWant – NumWish (0.940)

NumWish – NumUserRatings (0.814)

To avoid duplicating information in the clustering process, we removed NumUserRatings and NumWish, keeping NumOwned and NumWant.
These retained features are more interpretable, more reliable, and represent the core popularity metrics of the games.
Removing the secondary and noisier variables prevents overweighting popularity and ensures balanced K-means clusters.

```python
# CASE 1 – All 23 features
case1_features = df.columns.tolist()

# CASE 2 – Numeric + popularity + rating (no binary categories)
case2_features = [
    'YearPublished', 'GameWeight', 'MinPlayers', 'MaxPlayers', 'ComAgeRec',
    'NumOwned', 'NumWant', 'ComMinPlaytime', 'ComMaxPlaytime', 'MfgAgeRec',
    'NumExpansions', 'NumImplementations', 'Rating'
]

# CASE 3 – Core features
case3_features = [
    'GameWeight',
    'NumOwned',
    'NumWant',
    'NumExpansions',
    'Rating'
]

print("Case 1:", len(case1_features))
print("Case 2:", len(case2_features))
print("Case 3:", len(case3_features))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import MinMaxScaler

def run_kmeans_for_case(df, features, k_range=range(2, 11), case_name="Case"):
    print(f"\n🔹 {case_name} – {len(features)} features\n")

    # Select
    X = df[features].copy()

    # Scale
    scaler = MinMaxScaler()
    X_scaled = scaler.fit_transform(X)

    sse = []
    sil = []

    for k in k_range:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        labels = kmeans.fit_predict(X_scaled)

        sse.append(kmeans.inertia_)
        sil.append(silhouette_score(X_scaled, labels))

    # Plot
    fig, axs = plt.subplots(1, 2, figsize=(12, 5))

    axs[0].plot(list(k_range), sse, marker='o')
    axs[0].set_title(f'Elbow – {case_name}')
    axs[0].set_xlabel('k')
    axs[0].set_ylabel('SSE')

    axs[1].plot(list(k_range), sil, marker='o')
    axs[1].set_title(f'Silhouette – {case_name}')
    axs[1].set_xlabel('k')
    axs[1].set_ylabel('Silhouette')

    plt.show()

    for k, sc in zip(k_range, sil):
        print(f"k={k} → silhouette={sc:.4f}")

    return X_scaled, sse, sil
```

```python
# CASE 1 – All features
X1_scaled, sse1, sil1 = run_kmeans_for_case(df, case1_features, case_name="Case 1 – All Features")

# CASE 2 – Numeric only
X2_scaled, sse2, sil2 = run_kmeans_for_case(df, case2_features, case_name="Case 2 – Numeric Only")

# CASE 3 – Core features
X3_scaled, sse3, sil3 = run_kmeans_for_case(df, case3_features, case_name="Case 3 – Core Features")
```

```python
final_features = ['NumOwned', 'NumWant', 'ComMinPlaytime', 'ComMaxPlaytime', 'Rating']

X_final = df[final_features].copy()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_final)
```

```python
from sklearn.cluster import KMeans

k = 3
kmeans_final = KMeans(n_clusters=k, random_state=42, n_init=10)
labels = kmeans_final.fit_predict(X_scaled)

df['Cluster'] = labels  # add cluster to original dataframe
```

```python
cluster_sizes = df['Cluster'].value_counts().sort_index()
print("\nCluster Sizes:\n", cluster_sizes)
```

```python
import pandas as pd

centroids = pd.DataFrame(
    scaler.inverse_transform(kmeans_final.cluster_centers_),
    columns=final_features
)

print("\nCluster Centroids (original scale):\n")
print(centroids)
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=labels, cmap='viridis', s=12)
plt.title("K-Means Clustering (k=3) – PCA Visualization")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar(label="Cluster")
plt.show()
```

```python
from sklearn.decomposition import PCA
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt

# 3D PCA
pca_3d = PCA(n_components=3)
X_pca_3d = pca_3d.fit_transform(X_scaled)

fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot
scatter = ax.scatter(
    X_pca_3d[:,0],
    X_pca_3d[:,1],
    X_pca_3d[:,2],
    c=labels,
    cmap='viridis',
    s=10
)

ax.set_title("K-Means Clustering (k=3) – 3D PCA Visualization")
ax.set_xlabel("PCA 1")
ax.set_ylabel("PCA 2")
ax.set_zlabel("PCA 3")

# Colorbar
cbar = plt.colorbar(scatter, ax=ax, shrink=0.5)
cbar.set_label("Cluster")

plt.show()
```

```python
# --- TEST EXTRA K VALUES ---
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

test_k_values = [2, 4, 5]

for k in test_k_values:
    kmeans_test = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_test = kmeans_test.fit_predict(X_scaled)

    sil = silhouette_score(X_scaled, labels_test)
    sse = kmeans_test.inertia_

    print(f"\n===== K = {k} =====")
    print(f"Silhouette Score: {sil:.4f}")
    print(f"SSE: {sse:.0f}")
```
